```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
root = Path(
    "/home/dbruwel/main/honours/pachner_graph_triangulations/data/results/sgd_models_dehydration/statistical_tests_nci/statistical_9000pflop_768_12/"
)

data = pd.read_csv(root / "scored_spheres_data.csv")
data.columns = ["sig", "nll"]

print(data['nll'].mean(), data['nll'].std())
# np.log(np.mean(np.exp(data["nll"] - data["nll"].mean())))
```

```python
from regina.engine import Triangulation3

vedgs = []
for sig in data["sig"]:
    t = Triangulation3.rehydrate(sig)
    edg = []
    for e in t.edges():
        edg.append(e.degree())

    vedg = np.var(edg)
    vedgs.append(vedg)
```
